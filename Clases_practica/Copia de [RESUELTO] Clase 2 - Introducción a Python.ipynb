### Ejercicio 1: Número par o impar
Escribe un programa que pida al usuario un número y verifique si es par o impar. Imprime el resultado.

```python
num = int(input('Ingresa un numero'))
if (num % 2) == 0:
    print('El numero ingresado es un numero par')
else:
    print('El numero ingresado es impar')
```

### Ejercicio 2: Suma de números

Escribe un programa que pida al usuario un número y siga pidiendo números hasta que el usuario ingrese 0. Al final, imprime la suma de todos los números ingresados.

```python
suma = 0
num_usuario = 1
while num_usuario != 0:
    num_usuario = int(input('Ingrese un numero '))
    suma = suma + num_usuario
print(f'El resultado final es: {suma}')
```

### Ejercicio 3: Palabras en jeringoso
Escribe un programa que reciba una lista de palabras y devuelva una nueva lista con cada palabra convertida a jeringoso.

*   Si la vocal es "a", se le añade "pa".
*   Si la vocal es "e", se le añade "pe".
*   Si la vocal es "i", se le añade "pi".
*   Si la vocal es "o", se le añade "po".
*   Si la vocal es "u", se le añade "pu".

Ejemplo: 'Gato' pasaría a ser 'Gapatopo'

Instrucciones:
*   Define una lista de palabras.
*   Usa un bucle for para recorrer cada palabra.
*   Para cada palabra, usa otro bucle for para convertirla a jeringoso.
*   Guarda las palabras convertidas en una nueva lista y devuelve el resultado.

```python
lista_palabras = ['Hola', 'Cierre', 'DICCIONARIO', 'Gato', 'Perro']
lista_vocales = ['a', 'e', 'i', 'o', 'u']
lista_palabra_jeringoso = []
for palabra in lista_palabras:
    palabra_minuscula = palabra.lower()
    palabra_final = ""
    for letra in palabra_minuscula:
        if letra not in lista_vocales:
            palabra_final = palabra_final + letra
        else:
            palabra_final = palabra_final + letra + 'p' + letra

    lista_palabra_jeringoso.append(palabra_final)
print(lista_palabra_jeringoso)
```

###Ejercicio 4: Número Primo
Crea una función llamada `es_primo(num)` que reciba un número entero y devuelva True si el número es primo y False en caso contrario. Imprime el resultado para varios números.

```python
def es_primo(num: int):
    for i in range(2,num):
        if num % i == 0:
            return False
    return True
```

###Ejercicio 5: Cadena sin vocales
Escribe una función `cadena_sin_vocales(cadena)` que reciba una cadena y devuelva la cadena pero sin las vocales. Prueba la función con diferentes cadenas.

```python
def cadena_sin_vocales(cadena: str):
    cadena_sin_vocales = ''
    for letra in cadena:
        if letra not in ['a', 'e', 'i', 'o', 'u']:
            cadena_sin_vocales += letra
    return cadena_sin_vocales
```

```python
cadena = 'Holaaaaa muuuoiouiouiy buenos dias'

print(cadena_sin_vocales(cadena))
```

### Ejercicio 6: Clase Rectangulo

**Objetivo:**

Crear una clase Rectangulo que permita calcular el área y el perímetro de un rectángulo.

**Instrucciones:**
1. Define la clase Rectangulo con los atributos ancho y alto.
2. Implementa los métodos calcular_area() y calcular_perimetro().
3. Crea una instancia de la clase y llama a ambos métodos.

```python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def calcular_area(self):
        return self.ancho * self.alto

    def calcular_perimetro(self):
        return 2 * (self.ancho + self.alto)

rectangulo = Rectangulo(ancho = 5, alto = 10)
rectangulo.calcular_perimetro()
```

### Ejercicio 7: Composición
**Objetivo:**

Crear una clase `Estudiante` que tenga un atributo `Curso` mediante composición.

**Instrucciones:**
1. Define la clase `Curso` con los atributos `nombre` y `duracion`.
2. Define la clase `Estudiante` con el atributo `nombre` y una instancia de `Curso`.
3. Crea una instancia de `Estudiante` y usa los métodos de `Curso`.

```python
class Curso:
    def __init__(self, nombre, duracion):
        self.nombre = nombre
        self.duracion = duracion
    def detalle_curso(self):
        return f"Curso: {self.nombre}, Duración: {self.duracion} horas."

class Estudiante:
    def __init__(self, nombre, curso):

        self.nombre = nombre
        self.curso = curso

    def mostrar_detalles(self):
        print(f"Estudiante: {self.nombre}")
        print(self.curso.detalle_curso())

curso_python = Curso('Python', 3)
estudiante_1 = Estudiante("Ana", curso_python)
```

```python
estudiante_1.mostrar_detalles()
```
