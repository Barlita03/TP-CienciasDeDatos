# Introducción a Numpy

Numpy es una biblioteca fundamental para realizar cálculos numéricos en Python. Nos permite crear y manipular arrays multidimensionales y realizar operaciones matemáticas de manera eficiente.

```python
# Importar la librería Numpy
import numpy as np
```

## Creación de Arrays
Podemos crear arrays de varias formas. Aquí mostramos ejemplos de arrays unidimensionales, bidimensionales, de ceros y con números aleatorios. Numpy también permite crear arrays de cualquier tamaño y con diferentes tipos de datos.

```python
# Crear un array unidimensional
array1D = np.array([1, 2, 3, 4, 5])
print("Array 1D:", array1D)
```

```python
# Crear un array bidimensional
array2D = np.array([[1, 2, 3], [4, 5, 6]])
print("Array 2D:\n", array2D)
```

```python
# Crear un array de ceros
array_zeros = np.zeros((3, 3))
print("Array de ceros:\n", array_zeros)
```

```python
# Crear un array de números aleatorios
array_random = np.random.rand(2, 3)
print("Array aleatorio:\n", array_random)
```

```python
equi_linspace = np.linspace(0, 1, 5)
print(equi_linspace)
```

```python
equi_arange = np.arange(0, 1, 0.1)
print(equi_arange)
```

## Operaciones Básicas con Arrays

## Propiedades de arrays

## Indexación y Slicing
Los arrays de Numpy soportan operaciones avanzadas de indexación y slicing, similares a las listas de Python, pero más poderosas.

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr[0], arr[1:4], arr[-1])
```

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print(matriz[0, 1], matriz[:, 2])
```

## Manipulación de formas

Numpy permite cambiar la forma de un array sin cambiar su contenido.

```python
arr = np.arange(6)
reshaped = arr.reshape((2, 3))
print(arr)
print(reshaped)
```

## Estadísticas con Numpy
Numpy facilita la realización de cálculos estadísticos como media, mediana, desviación estándar y suma de los elementos de un array.

```python
array1D = np.array([1, 2, 3, 4, 5])
```

```python
# Calcular la media de un array
media = np.mean(array1D)
print("Media:", media)

# Calcular la mediana
mediana = np.median(array1D)
print("Mediana:", mediana)

# Calcular la desviación estándar
desviacion_estandar = np.std(array1D)
print("Desviación Estándar:", desviacion_estandar)

# Calcular la suma de los elementos
suma_total = np.sum(array1D)
print("Suma de los elementos:", suma_total)
```

# Introducción a Pandas

Pandas es una librería poderosa para el manejo y análisis de datos. Facilita la manipulación de grandes volúmenes de datos de forma eficiente, a través de estructuras como Series y DataFrames.

```python
# Importar la librería Pandas
import pandas as pd
```

## Series de Pandas

Las Series en Pandas son estructuras unidimensionales similares a los arrays de Numpy, pero con etiquetas (índices) para los datos.

```python
# Crear una Serie
serie = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Serie:\n", serie)
```

```python
# Acceder a un valor por índice
print("Valor en el índice 'c':", serie[['c', 'd']])
```

## DataFrames de Pandas
Los DataFrames son estructuras bidimensionales donde los datos se organizan en filas y columnas, similares a las tablas de bases de datos o Excel. Permiten acceder tanto a filas como a columnas de forma fácil.

```python
# Crear un DataFrame a partir de un diccionario
data = {'Nombre': ['Ana', 'Luis', 'Pedro'], 'Edad': [25, 30, 22], 'Salario': [50000, 60000, 45000]}
df = pd.DataFrame(data)
print("DataFrame:\n", df)
```

```python
# Acceder a una columna
print("Columna 'Edad':\n", df[['Salario', 'Edad']])
```

```python
# Acceder a una fila
print("Fila con índice 1:\n", df.iloc[0:4])
```

### Filtrado de Datos
Podemos filtrar los datos en función de las columnas. En este caso, hemos filtrado los pasajeros que sobrevivieron y aquellos que viajaban en primera clase.

```python
# Filtrar pasajeros que sobrevivieron
sobrevivientes = titanic_df[titanic_df['survived'] == 1]
print("Número de sobrevivientes:", len(sobrevivientes))
```

### Agregación de Datos
Podemos agrupar los datos por diferentes categorías para obtener estadísticas más específicas. Aquí agrupamos por clase y sexo para calcular la tarifa media y la tasa de supervivencia respectivamente.

```python
 # Agrupar por clase y calcular la media de la tarifa pagada
tarifa_media_por_clase = titanic_df.groupby('pclass')['fare'].mean()
print("Tarifa media por clase:\n", tarifa_media_por_clase)
```

### Joins con Pandas
Los joins se utilizan para combinar dos DataFrames en uno solo, basándose en una columna común. Veamos cómo cambia el resultado según el parámetro `how` del método `merge` (`inner`, `left`, `right` y `outer`).


```python
import pandas as pd

# Crear primer DataFrame con información de personas
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'nombre': ['Ana', 'Carlos', 'Daniel', 'Eva', 'Fernando'],
    'edad': [28, 35, 42, 23, 31]
})

# Crear segundo DataFrame con información adicional
df2 = pd.DataFrame({
    'id': [3, 4, 5, 6, 7],
    'ciudad': ['Madrid', 'Barcelona', 'Sevilla', 'Valencia', 'Bilbao'],
    'ocupacion': ['Ingeniero', 'Profesor', 'Doctor', 'Artista', 'Abogado']
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python

# Hacer un inner join (solo coincidencias en ambos DataFrames)
inner_join = pd.merge(df1, df2, on='id', how='inner')
print("\nInner Join (solo personas con ID en ambos DataFrames):")
print(inner_join)
```

```python
# Hacer un left join (todas las filas del primer DataFrame y coincidencias del segundo)
left_join = pd.merge(df1, df2, on='id', how='left')
print("\nLeft Join (todas las personas de df1, rellenando con NaN los que no están en df2):")
print(left_join)
```

```python

# Hacer un right join (todas las filas del segundo DataFrame y coincidencias del primero)
right_join = pd.merge(df1, df2, on='id', how='right')
print("\nRight Join (todas las personas de df2, rellenando con NaN los que no están en df1):")
print(right_join)
```

```python

# Hacer un outer join (todas las filas de ambos DataFrames, sin importar coincidencias)
outer_join = pd.merge(df1, df2, on='id', how='outer')
print("\nOuter Join (todas las personas, sin importar si están en uno o ambos DataFrames):")
print(outer_join)
```

### Visualización de la distribución de las variables
Para visualizar la distribución de las variables numéricas en el dataset, utilizamos histogramas. Estos nos permiten ver la forma en que se distribuyen las observaciones, así como la presencia de sesgos o valores extremos.


```python
# Visualización de la distribución de las variables
plt.figure(figsize=(12, 6))
df.iloc[:, :-1].hist(bins=20, edgecolor='black', alpha=0.7, layout=(2, 2), figsize=(12, 8))
plt.suptitle('Distribución de las Variables Numéricas')
plt.show()
```

### Gráfico de dispersión (Scatter plot) entre las variables
El gráfico de dispersión nos permite visualizar las relaciones entre pares de variables numéricas. Al utilizar el parámetro `hue='species'`, podemos colorear las observaciones según la especie de flor, lo cual nos ayuda a identificar cómo las distintas especies se agrupan en función de las variables. Este tipo de visualización es muy útil para detectar patrones entre características y ver posibles separaciones entre clases.

El argumento `diag_kind='hist'` indica que en la diagonal se mostrará el histograma de cada variable. Así podemos ver las distribuciones de forma individual, además de las relaciones entre pares de variables.


```python
import warnings

# Suprimir la advertencia de Pandas
warnings.filterwarnings("ignore", message="When grouping with a length-1 list-like")


# Gráfico de dispersión (Scatter plot) entre las variables
sns.pairplot(df, hue='species', palette='viridis', diag_kind='hist')
plt.suptitle('Relación entre las Variables', y=1.02)
plt.show()
```

### Verificación de valores atípicos con boxplots
Los diagramas de caja o *boxplots* nos permiten visualizar la distribución de las variables numéricas y, a la vez, identificar posibles valores atípicos (outliers). La caja central muestra el rango intercuartílico (IQR), y los "bigotes" muestran la extensión de los datos, más allá de los cuales los puntos se consideran atípicos.

Al usar `hue='species'`, creamos un gráfico de cajas por especie, lo cual facilita la comparación de la distribución de cada variable entre las diferentes especies de flores. Esta visualización es particularmente útil para evaluar si una especie tiene una mayor dispersión en ciertas características.


```python
warnings.filterwarnings("ignore", message="SeriesGroupBy.grouper is deprecated")
# Verificación de valores atípicos con boxplots
plt.figure(figsize=(12, 8))
for i, col in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i+1)
    sns.boxplot(x='species',hue = 'species', y=col, data=df, palette='viridis')
    plt.title(f'Boxplot de {col}')
plt.tight_layout()
plt.show()
```

### Gráfico de violín por especie
El gráfico de violín combina las propiedades del boxplot con una estimación de la densidad de la distribución de los datos. Asignamos tanto `x` como `hue` a la columna `species`, lo cual permite aplicar la paleta de colores correctamente. Esto nos permite visualizar no solo el rango y los valores atípicos de cada variable, sino también la forma de su distribución en diferentes especies. Es útil para observar cómo se distribuyen las características entre las diferentes clases de flores y detectar patrones o sobreposiciones en las distribuciones de las variables.


```python
# Gráfico de violín para cada variable numérica por especie
plt.figure(figsize=(12, 8))
for i, col in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i+1)
    sns.violinplot(x='species',hue = 'species', legend = False, y=col, data=df, palette='viridis')
    plt.title(f'Gráfico de Violín para {col}')
plt.tight_layout()
plt.show()
```

### Heatmap de la correlación
La matriz de correlación mide la relación entre pares de variables numéricas. Un valor cercano a 1 indica una alta correlación positiva, mientras que un valor cercano a -1 indica una alta correlación negativa. En este gráfico de calor (heatmap), los valores de correlación se representan por colores, con tonos más cálidos (rojos) indicando correlaciones positivas y tonos más fríos (azules) indicando correlaciones negativas.

Este gráfico nos ayuda a identificar rápidamente qué variables están relacionadas entre sí, lo cual puede ser útil para reducir la dimensionalidad o evitar la multicolinealidad en modelos de machine learning.


```python
# 4. Visualizaciones
## Heatmap de la correlación
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Matriz de Correlación')
plt.show()
```
